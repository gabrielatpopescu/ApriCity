💢 Tipul de grafic pe care îl folosim depinde de tipul variabilei

1. Un grafic de tip **linie** (line chart) este folosit pentru a arăta relația dintre două variabile **continue**.
2. Un grafic de tip **coloană** (column chart/**bar chart**) este folosit atunci când variabila independentă (de cele mai multe ori cea de pe axa Ox) este o variabilă **discretă** sau **calitativă**.

# Tipuri comune de grafice

- **Line chart** -> de obicei prezintă schimbări în timp
- **Bar chart** -> lucruri similare dintr-o singură categorie
- **Double bar chart** -> compară date
- **Histogram** -> tip de bar chart; compară intervale de date dintr-o singură categorie
- **Stemplot** -> Asemănătoare Histogramelor, cu mențiunea că afișează fiecare punct
- **Pie chart** -> Arată părți dintr-un întreg (de obicei pentru afișarea procentelor)
- **Scatter plot** -> Puncte care nu sunt conectate, de obicei pentru a arăta o corelație (va fi discutat mai pe larg în laboratoarele viitoare)
- **Time plot** -> Arată modul în care o variabilă evoluează în timp
💢 Convenție: timpul pe Ox, valorile variabilei pe Oy
- **Box and whisker** -> Prezintă variația datelor

# Un grafic bun:

1. Obligă la minimul de efort din partea cititorului (Legende vs labeluri)
2. Maximizează informația (Nr de ore petrecute în fața ecranului vs Nr de ore pe zi petrecute în fața ecranului)
3. Minimizează informația afișată (Liniile de grid: doar dacă sunt absolut necesare pentru a citi anumite valori)
4. Folosește bune practici (origine în (0,0), scale liniare, creștere de la stânga la dreapta etc.)
5. Evită ambiguitatea (axele, diviziunile, originea sunt vizibile).

# Checklist pentru grafice bune

1. Sunt ambele axe de coordonate afișate și adnotate?
2. Sunt adnotările axelor concise, dar suficient de clare?
3. Sunt scalele si diviziunile afișate pe ambele axe?
4. Sunt afișate minimul și maximul pe axe pentru a putea prezenta maximul de informație?
5. Este numărul de curbe rezonabil? (Ex: Un line chart nu prea are mai mult de 6 curbe)
6. Folosesc toate graficele aceeași scală? (Mai multe scale pe același chart -> redundanță)
7. Se poate elimina vreo curbă fără a reduce informația?
8. Este fiecare curbă dintr-un line chart adnotată?
9. Este fiecare celulă dintr-un bar chart adnotată?
10. Sunt toate simbolurile dintr-un grafic însoțite de explicații textuale?
11. În cazul în care curbele se intersectează, au pattern-uri diferite, pentru a evita confuzia?
12. Sunt unitățile de măsură indicate?
13. Scala orizontală are creștere de la stânga la dreapta?
14. Scala verticală are creștere de jos în sus?
15. Ajută liniile de grid la citirea graficelor? (Dacă nu, nu le afișați)
16. Sunt scalele continue? (Ar trebui evitat sau menționat separat)
17. Ordinea barelor dintr-un bar chart este sistematică? (alfabetic, temporal, best-to-worst...orice, dar nu random)
18. Dacă axa verticală reprezintă o cantitate probabilistică, este afișat și intervalul de încredere? (va fi discutat mai târziu)
19. Există curbe, simboluri, text în figură care să poată fi eliminat fără a afecta informația?
20. Am adăugat un titlu?
21. Este titlul concis, dar suficient de clar?
22. Variabilele folosite explică mai bine fenomenul decât altele alternative?
23. Graficul expune mesajul pe care mi l-am propus?

# Scatter & Line Plots

```python
# Generate some random values

M = 10
N = 5

X = np.random.random(size = 100)
y1 = M * X + N
y2 = y1 + np.random.normal(size = 100)
```

```python
plt.plot(X, y1)
plt.xlim(-0.2, 1.2)
plt.ylim(0, 20)

plt.xlabel("X")
plt.ylabel("Y1")

plt.title("Simple Scatter Plot for Generated Data")
plt.gcf().set_size_inches(14,7)
```

```python
plt.scatter(X, y2)
plt.xlim(-0.2, 1.2)
plt.ylim(0, 20)

plt.xlabel("X")
plt.ylabel("Y2")

plt.title("Simple Scatter Plot for Generated Data")
```

💢 Matplotlib **markers** full list: https://matplotlib.org/api/markers_api.html

```python
plt.scatter(X, y2, label='y2', marker = '<')
plt.plot(X, y1, color='r', label='y1', linestyle='dashed', linewidth=1)

plt.xlim(-0.2, 1.2)
plt.ylim(0, 20)

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Simple Scatter Plot for Generated Data")
plt.legend()
plt.gcf().set_size_inches(14, 7)
```

# Bar Plots

```python
languages = ['C', 'C++', 'Java', 'Python', 'PHP']
students = [10, 12, 35, 32, 5]

ax = plt.subplot()
ax.bar(languages, students)

plt.xlabel('Programming Language')
plt.ylabel('Number of favourites among students')
plt.gcf().set_size_inches(14, 7)
```

💢Numpy arange: https://numpy.org/doc/stable/reference/generated/numpy.arange.html

```python
data = [
    [30, 25, 50, 20],
    [40, 23, 51, 17],
    [35, 22, 45, 19]
]

X = np.arange(4)
ax = plt.subplot()

ax.bar(X + 0.00, data[0], color = 'b', width = 0.25, label = 'CTI')
ax.bar(X + 0.25, data[1], color = 'g', width = 0.25, label = 'IS')
ax.bar(X + 0.50, data[2], color = 'r', width = 0.25, label = 'Other')

ax.set_xticks(X + 0.25)
ax.set_xticklabels(X + 2017)

plt.xlabel("Year")
plt.ylabel("Number of students")
plt.title("Number of students for each department over the years.")
ax.legend()
plt.gcf().set_size_inches(14, 7)
```

# Box Plots

💢https://builtin.com/data-science/boxplot

```python
data = np.random.normal(size = 100, scale = 3.0, loc = 1.0)

ax = plt.subplot()
ax.boxplot([data, data ** 2, data * 2])

plt.title("Box plots for skewed normal distributions.")
plt.gcf().set_size_inches(14, 7)
```
